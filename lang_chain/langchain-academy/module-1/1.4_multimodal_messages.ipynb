# Text input

```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langgraph.checkpoint.memory import InMemorySaver

model = init_chat_model("ollama:nemotron3:33b", temperature=0.15)

agent = create_agent(
    model=model,
    checkpointer=InMemorySaver(),
    system_prompt="You are a science fiction writer, create a capital city at the users request."
)
```

```python
import json
from langchain.messages import HumanMessage

question = HumanMessage(content=[
    {"type": "text", "text": "What is the capital of The Moon?"}
])
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config=config,
 )

last_message = response['messages'][-1]
print(last_message.content)
print(json.dumps(last_message.usage_metadata, indent=4))
print(json.dumps(last_message.response_metadata, indent=4))
```

# Image Input

```python

from base64 import b64encode
from langchain.messages import HumanMessage

img_path="resources/moon.png"
img_b64 = b64encode(open(img_path, "rb").read()).decode("utf-8")

multimodal_question = HumanMessage(content=[
    {"type": "text", "text": "Tell me about this capital"},
    {"type": "image", "base64": img_b64, "mime_type": "image/png"}
])
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [multimodal_question]},
    config=config
)

last_message = response['messages'][-1]
print(last_message.content)
print(json.dumps(last_message.usage_metadata, indent=4))
print(json.dumps(last_message.response_metadata, indent=4))
```

# Audio Input

```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langgraph.checkpoint.memory import InMemorySaver

model = init_chat_model("ollama:nemotron3:33b", temperature=0.15)

agent = create_agent(
    model=model,
    checkpointer=InMemorySaver()
)
# nemotron3:33b
```

```python
import base64
import logging
from pathlib import Path

from langchain.messages import HumanMessage

logger = logging.getLogger(__name__)

video_path = Path("resources/3 MAJOR Driving Habits That Extend Toyota Hybrid Battery Life!.webm")
if not video_path.exists():
    raise FileNotFoundError(f"Video file not found: {video_path}")

try:
    with video_path.open("rb") as file_obj:
        video_b64 = base64.b64encode(file_obj.read()).decode("utf-8")
except OSError:
    logger.exception("Failed reading video file at path %s", video_path)
    raise

multimodal_question = HumanMessage(
    content=[
        {"type": "text", "text": "What is the correct way to stop a car?"},
        {"type": "video", "base64": video_b64, "mime_type": "video/webm"},
    ]
)

# nemotron3:33b
config = {"configurable": {"thread_id": "1"}}
response = agent.invoke({"messages": [multimodal_question]}, config=config)
last_message = response["messages"][-1]
last_message.content
```
